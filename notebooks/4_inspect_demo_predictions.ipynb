## Inspecting predictions from the demo

Inference has already been run on a validation set of 3,000 doctor–patient conversations, using the same system prompt as in training. Here we simply load and display those results. **Re-run this cell to see a new random example.**


Inference has already been run on a validation set of 3,000 doctor–patient conversations. Here we simply load and display those results. **Re-run the cell below to see a new random example.**

Each record in the `data/predictions.json` file contains:

| Field | Description |
|---|---|
| `conversation` | The doctor–patient dialogue |
| `reference` | Ground-truth reference answer |
| `base_4b_response` | MedGemma-4B original model response|
| `finetuned_response` | MedGemma-4B fine-tuned model response|
| `base_27b_response` | MedGemma-27B original model response|

```python
# Run this cell again to get a new random example
import json

pred_file = f"/scratch/project_462001520/data/predictions.json"

with open(pred_file, encoding="utf-8") as f:
    records = json.load(f)

def format_conversation(text: str) -> str:
    """Add a blank line after each Doctor/Patient turn for readable Markdown."""
    return text.replace("Doctor:", "\n\n**Doctor:**").replace("Patient:", "\n\n**Patient:**")

from IPython.display import Markdown, display
import random

def show_random_example(records: list) -> None:
    record = random.choice(records)
    md = f"""
---
## Random Example

### 💬 Conversation
{format_conversation(record["conversation"])}

### ✅ Reference (ground truth)
{record["reference"]}

### 🔵 Model 1: Medgemma-4B Base
{record["base_4b_response"]}

### 🟠 Model 2: Medgemma-4B Finetuned
{record["finetuned_response"]}

### 🟢 Model 3: Medgemma-27B Base
{record["base_27b_response"]}
"""
    display(Markdown(md))

show_random_example(records)
```
